## 3 - 梯度检查是如何工作的？
反向传播计算梯度 $\frac{\partial J}{\partial \theta}$，其中 $\theta$ 表示模型的参数。 $J$ 是使用前向传播和你的损失函数计算的。

因为前向传播相对容易实现，所以你确信你的计算是正确的，因此你几乎 100% 确定你正确计算了代价 $J$。因此，你可以使用计算 $J$ 的代码来验证计算 $\frac{\partial J}{\partial \theta}$ 的代码。

我们回顾一下导数（或梯度）的定义：$$ \frac{\partial J}{\partial \theta} = \lim_{\varepsilon \to 0} \frac{J(\theta + \varepsilon) - J(\theta - \varepsilon)}{2 \varepsilon} \tag{1}$$

如果你不熟悉“$\displaystyle \lim_{\varepsilon \to 0}$”表示法，它只是“当 $\varepsilon$ 非常非常小的时候”的一种说法。

你知道以下内容：

$\frac{\partial J}{\partial \theta}$ 是你想要确保计算正确的值。
你可以计算 $J(\theta + \varepsilon)$ 和 $J(\theta - \varepsilon)$（在 $\theta$ 是实数的情况下），因为你确信 $J$ 的实现是正确的。
让我们使用方程（1）和 $\varepsilon$ 的一个小值来说服你的 CEO，你计算 $\frac{\partial J}{\partial \theta}$ 的代码是正确的！

让我们看看 `forward_propagation` 和 `backward_propagation` 的实现。 

以下是为 `forward_propagation_n` 提供的代码。请注意，名称中的 `n` 意味着它适用于 n 维

````python
def forward_propagation_n(X, Y, parameters):
    """
    实现图 3 中所示的前向传播（并计算代价）。
    
    参数：
    X -- 训练集 m 个示例
    Y -- m 个示例的标签 
    parameters -- 包含参数"W1"、"b1"、"W2"、"b2"、"W3"、"b3"的python字典：
                    W1 -- 形状 (5, 4) 的权重矩阵
                    b1 -- 形状为 (5, 1) 的偏置向量
                    W2 -- 形状 (3, 5) 的权重矩阵
                    b2 -- 形状为 (3, 1) 的偏置向量
                    W3 -- 形状 (1, 3) 的权重矩阵
                    b3 -- 形状为 (1, 1) 的偏置向量
    
    返回：
    cost -- 代价函数（m 个示例的logistic代价）
    cache -- 中间值 (Z1, A1, W1, b1, Z2, A2, W2, b2, Z3, A3, W3, b3) 的元组

    """
    
    # 检索参数
    m = X.shape[1]
    W1 = parameters["W1"]
    b1 = parameters["b1"]
    W2 = parameters["W2"]
    b2 = parameters["b2"]
    W3 = parameters["W3"]
    b3 = parameters["b3"]

    # 线性 -> ReLU -> 线性 -> ReLU -> 线性 -> sigmoid
    Z1 = np.dot(W1, X) + b1
    A1 = relu(Z1)
    Z2 = np.dot(W2, A1) + b2
    A2 = relu(Z2)
    Z3 = np.dot(W3, A2) + b3
    A3 = sigmoid(Z3)

    # 代价
    log_probs = np.multiply(-np.log(A3),Y) + np.multiply(-np.log(1 - A3), 1 - Y)
    cost = 1. / m * np.sum(log_probs)
    
    cache = (Z1, A1, W1, b1, Z2, A2, W2, b2, Z3, A3, W3, b3)
    
    return cost, cache
````

现在，让我们看一下反向传播的代码。

以下是为 `backward_propagation_n` 提供的代码。请注意，名称中的 `n` 意味着它适用于 n 维

````python
def backward_propagation_n(X, Y, cache):
    """
    实现图 2 中所示的反向传播。
    
    参数：
    X -- 输入数据点，形状（输入大小，1）
    Y -- 真正的 "label"
    cache -- 缓存 forward_propagation_n() 的输出
    
    返回：
    gradients -- 字典，包含代价相对于各参数、激活值及预激活值变量的梯度。
    """
    
    m = X.shape[1]
    (Z1, A1, W1, b1, Z2, A2, W2, b2, Z3, A3, W3, b3) = cache
    
    dZ3 = A3 - Y
    dW3 = 1. / m * np.dot(dZ3, A2.T)
    db3 = 1. / m * np.sum(dZ3, axis=1, keepdims=True)
    
    dA2 = np.dot(W3.T, dZ3)
    dZ2 = np.multiply(dA2, np.int64(A2 > 0))
    dW2 = 1. / m * np.dot(dZ2, A1.T) * 2
    db2 = 1. / m * np.sum(dZ2, axis=1, keepdims=True)
    
    dA1 = np.dot(W2.T, dZ2)
    dZ1 = np.multiply(dA1, np.int64(A1 > 0))
    dW1 = 1. / m * np.dot(dZ1, X.T)
    db1 = 4. / m * np.sum(dZ1, axis=1, keepdims=True)
    
    gradients = {"dZ3": dZ3, "dW3": dW3, "db3": db3,
                 "dA2": dA2, "dZ2": dZ2, "dW2": dW2, "db2": db2,
                 "dA1": dA1, "dZ1": dZ1, "dW1": dW1, "db1": db1}
    
    return gradients
````

**梯度检查是如何工作的？**。

与第 3 节和第 4 节一样，你想要将“gradapprox”与反向传播计算的梯度进行比较。公式依然是：

$$ \frac{\partial J}{\partial \theta} = \lim_{\varepsilon \to 0} \frac{J(\theta + \varepsilon) - J(\theta - \varepsilon)}{2 \varepsilon} \tag{1}$$

然而，$\theta$ 不再是标量。它是一本叫做“参数”的字典。 函数“`dictionary_to_vector()`”已为你实现。它将“参数”字典转换为一个称为“值”的向量，该向量是通过将所有参数（W1、b1、W2、b2、W3、b3）重塑为向量并将它们连接起来而获得的。

函数的逆是“`vector_to_dictionary`”，它输出回“参数”字典。

<img src="images/dictionary_to_vector.png" style="width:600px;height:400px;">
<caption><center><font color='purple'><b>图2</b>：dictionary_to_vector()和vector_to_dictionary()。你将在 gradient_check_n()</font></center></caption> 中需要这些函数

“梯度”字典也已使用 gradients_to_vector() 转换为向量“grad”，因此你无需担心这一点。

现在，对于向量中的每个参数，你将应用与 gradient_check 练习相同的过程。你将把每个梯度近似值存储在向量 `gradapprox` 中。如果检查按预期进行，则该近似值中的每个值都必须与 `grad` 向量中存储的真实梯度值匹配。 

注意 `grad` 的计算方法是使用函数 `gradients_to_vector`，它使用梯度的输出 `backward_propagation_n` 函数.

- 梯度检查很慢！用 $\frac{\partial J}{\partial \theta} \approx  \frac{J(\theta + \varepsilon) - J(\theta - \varepsilon)}{2 \varepsilon}$ 近似梯度的计算代价很高。因此，我们不会在训练期间的每次迭代中都运行梯度检查。只需几次即可检查梯度是否正确。
