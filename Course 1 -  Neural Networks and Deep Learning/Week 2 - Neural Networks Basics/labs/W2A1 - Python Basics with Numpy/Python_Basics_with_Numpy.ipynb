### 1.4 - 按行归一化

我们在机器学习和深度学习中使用的另一种常见技术是将我们的数据归一化。它通常会带来更好的性能，因为梯度下降在归一化之后收敛得更快。在这里，归一化的意思是将 x 更改为 $ \frac{x}{\| x\|} $（将 x 的每个行向量除以其范数）。

例如，如果

$$
x = \begin{bmatrix}
        0 & 3 & 4 \\
        2 & 6 & 4 \\
\end{bmatrix}\tag{3}
$$

 
那么

$$
\| x\| = \text{np.linalg.norm(x, axis=1, keepdims=True)} = \begin{bmatrix}
    5 \\
    \sqrt{56} \\
\end{bmatrix}\tag{4}
$$


和

$$
x\_normalized = \frac{x}{\| x\|} = \begin{bmatrix}
    0 & \frac{3}{5} & \frac{4}{5} \\
    \frac{2}{\sqrt{56}} & \frac{6}{\sqrt{56}} & \frac{4}{\sqrt{56}} \\
\end{bmatrix}\tag{5}
$$

请注意，你可以划分不同大小的矩阵，并且效果很好：这称为广播，你将在第 5 部分中了解它。

使用 `keepdims=True`，结果将根据原始 x 正确广播。

`axis=1` 意味着你将以行方式获得范数。如果你需要按列方式进行范数，则需要设置 `axis=0`。 

numpy.linalg.norm 还有另一个参数 `ord`，我们在其中指定要完成的归一化的类型（在下面的练习中，你将执行 2-范数）。要熟悉归一化的类型，你可以访问[numpy.linalg.norm](https://numpy.org/doc/stable/reference/generated/numpy.linalg.norm.html)

- $\text{for } x \in \mathbb{R}^{1\times n} \text{,     }$

$$

## 2 - 向量化


在深度学习中，你要处理非常大的数据集。因此，非计算最优的函数可能成为算法中的巨大瓶颈，并可能导致模型需要很长时间才能运行。为了确保你的代码具有计算效率，你将使用向量化。例如，尝试区分点/外/元素乘积的以下实现之间的区别。

````python
import time

x1 = [9, 2, 5, 0, 0, 7, 5, 0, 0, 0, 9, 2, 5, 0, 0]
x2 = [9, 2, 2, 9, 0, 9, 2, 5, 0, 0, 9, 2, 5, 0, 0]

### 向量实现的经典点积###
tic = time.process_time()
dot = 0

for i in range(len(x1)):
    dot += x1[i] * x2[i]
toc = time.process_time()
print ("dot = " + str(dot) + "\n ----- Computation time = " + str(1000 * (toc - tic)) + "ms")

### 经典外积实现 ###
tic = time.process_time()
outer = np.zeros((len(x1), len(x2))) # 我们创建一个只有零的 len(x1)*len(x2) 矩阵

for i in range(len(x1)):
    for j in range(len(x2)):
        outer[i,j] = x1[i] * x2[j]
toc = time.process_time()
print ("outer = " + str(outer) + "\n ----- Computation time = " + str(1000 * (toc - tic)) + "ms")

### 经典的逐元素实现 ###
tic = time.process_time()
mul = np.zeros(len(x1))

for i in range(len(x1)):
    mul[i] = x1[i] * x2[i]
toc = time.process_time()
print ("elementwise multiplication = " + str(mul) + "\n ----- Computation time = " + str(1000 * (toc - tic)) + "ms")

### 经典通用点积实现 ###
W = np.random.rand(3,len(x1)) # 随机3*len(x1) NumPy数组
tic = time.process_time()
gdot = np.zeros(W.shape[0])

for i in range(W.shape[0]):
    for j in range(len(x1)):
        gdot[i] += W[i,j] * x1[j]
toc = time.process_time()
print ("gdot = " + str(gdot) + "\n ----- Computation time = " + str(1000 * (toc - tic)) + "ms")
````

````python
x1 = [9, 2, 5, 0, 0, 7, 5, 0, 0, 0, 9, 2, 5, 0, 0]
x2 = [9, 2, 2, 9, 0, 9, 2, 5, 0, 0, 9, 2, 5, 0, 0]

### 向量化的向量### 的点积
tic = time.process_time()
dot = np.dot(x1,x2)
toc = time.process_time()
print ("dot = " + str(dot) + "\n ----- Computation time = " + str(1000 * (toc - tic)) + "ms")

### 向量化的外部产品 ###
tic = time.process_time()
outer = np.outer(x1,x2)
toc = time.process_time()
print ("outer = " + str(outer) + "\n ----- Computation time = " + str(1000 * (toc - tic)) + "ms")

### 向量化的元素乘法###
tic = time.process_time()
mul = np.multiply(x1,x2)
toc = time.process_time()
print ("elementwise multiplication = " + str(mul) + "\n ----- Computation time = " + str(1000*(toc - tic)) + "ms")

### 向量化的一般点积###
tic = time.process_time()
dot = np.dot(W,x1)
toc = time.process_time()
print ("gdot = " + str(dot) + "\n ----- Computation time = " + str(1000 * (toc - tic)) + "ms")
````

正如你可能已经注意到的，向量化的实现更加干净和高效。对于更大的向量/矩阵，运行时间的差异变得更大。 

**注意** `np.dot()` 执行矩阵-矩阵或矩阵-向量乘法。这与 `np.multiply()` 和 `*` 运算符（相当于 Matlab/Octave 中的 `.*`）不同，后者执行逐元素乘法。
