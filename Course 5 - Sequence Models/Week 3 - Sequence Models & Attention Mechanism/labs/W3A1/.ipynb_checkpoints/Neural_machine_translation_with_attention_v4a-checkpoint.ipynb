以下是你可能会注意到的模型的一些属性： 

#### 注意力机制两侧的 LSTM：注意力前与注意力后
- 模型包含两个独立的 LSTM（见左图）：注意力前 LSTM 和注意力后 LSTM。
- *注意力前* Bi-LSTM 位于图的底部，是一个双向 LSTM，在注意力机制*之前*运行。
    - 注意力机制如左图中间所示。
    - 注意力前 Bi-LSTM 经过 $T_x$ 时间步
- *注意力后* LSTM 位于图的上方，在注意力机制*之后*运行。
    - 注意力后 LSTM 经历 $T_y$ 时间步。 

- 注意力后 LSTM 将隐藏状态、$s^{\langle t \rangle}$ 和单元状态 $c^{\langle t \rangle}$ 从一个时间步传递到下一时间步。 

#### LSTM 同时具有隐藏状态和单元状态
* 在课程视频中，注意力后的序列模型只使用了基本 RNN。
    * 这意味着 RNN 捕获的状态仅输出隐藏状态 $s^{\langle t\rangle}$。 
* 在此作业中，我们使用 LSTM 而不是基本的 RNN。
    * 所以 LSTM 既有隐藏状态 $s^{\langle t\rangle}$ 又有单元状态 $c^{\langle t\rangle}$。 

#### 每个时间步不使用前一个时间步的预测
* 与课程前面的文本生成示例不同，在此模型中，时间 $t$ 的注意力后 LSTM 不会将前一个时间步的预测 $y^{\langle t-1 \rangle}$ 作为输入。
* t 时刻的注意力后 LSTM 仅将隐藏状态 $s^{\langle t\rangle}$ 和单元状态 $c^{\langle t\rangle}$ 作为输入。 
* 我们以这种方式设计模型是因为与语言生成（相邻字符高度相关）不同，YYYY-MM-DD 日期中的前一个字符和下一个字符之间没有那么强的依赖性。

#### 拼接注意力前 LSTM 的正向与反向隐藏状态
- $\overrightarrow{a}^{\langle t \rangle}$：注意力前 LSTM 沿正向处理序列时的隐藏状态。
- $\overleftarrow{a}^{\langle t \rangle}$：注意力前 LSTM 沿反向处理序列时的隐藏状态。
- $a^{\langle t \rangle} = [\overrightarrow{a}^{\langle t \rangle}, \overleftarrow{a}^{\langle t \rangle}]$：注意力前 Bi-LSTM 的前向 $\overrightarrow{a}^{\langle t \rangle}$ 和后向 $\overleftarrow{a}^{\langle t \rangle}$ 的激活值的拼接。 

#### 将“能量”$e^{\langle t, t' \rangle}$ 计算为 $s^{\langle t-1 \rangle}$ 和 $a^{\langle t' \rangle}$ 的函数
- 回想一下“注意力模型”课程视频中，6:45 到 8:16 时，“e”的定义为 $s^{\langle t-1 \rangle}$ 和 $a^{\langle t \rangle}$ 的函数。
    - “e”被称为“能量”变量。
    - $s^{\langle t-1 \rangle}$ 是注意力后 LSTM 的隐藏状态
    - $a^{\langle t' \rangle}$ 是注意力前 LSTM 的隐藏状态。
    - $s^{\langle t-1 \rangle}$ 和 $a^{\langle t \rangle}$ 被输入一个简单的神经网络，该网络学习函数并输出 $e^{\langle t, t' \rangle}$。
    - 然后在计算$y^{\langle t \rangle}$应该对$a^{\langle t' \rangle}$给予的关注度$\alpha^{\langle t, t' \rangle}$时使用$e^{\langle t, t' \rangle}$。

- 图1右图使用`RepeatVector`节点复制$s^{\langle t-1 \rangle}$的值$T_x$次。
- 然后它使用 `Concatenation` 连接 $s^{\langle t-1 \rangle}$ 和 $a^{\langle t \rangle}$。
- 将 $s^{\langle t-1 \rangle}$ 与 $a^{\langle t \rangle}$ 拼接后输入 Dense 层，计算 $e^{\langle t, t' \rangle}$。
- $e^{\langle t, t' \rangle}$ 通过 softmax 来计算 $\alpha^{\langle t, t' \rangle}$。
- 请注意，该图没有明确显示变量 $e^{\langle t, t' \rangle}$，但在图 1 右半部分的图中，$e^{\langle t, t' \rangle}$ 位于 Dense 层上方、softmax 层下方。
- 下面我们将解释如何在Keras中使用`RepeatVector`和`Concatenation`。 

#### 实现细节
   
让我们实现这个神经翻译器。你将首先实现两个函数：`one_step_attention()` 和 `model()`。

#### one_step_attention
* one_step_attention 在时间步 $t$ 的输入为：
    - $[a^{<1>},a^{<2>}, ..., a^{<T_x>}]$：注意力前 Bi-LSTM 的所有隐藏状态。
    - $s^{<t-1>}$：注意力后LSTM的前一个隐藏状态 
* one_step_attention 计算：
    - $[\alpha^{<t,1>},\alpha^{<t,2>}, ..., \alpha^{<t,T_x>}]$：注意力权重
    - $context^{ \langle t \rangle }$：上下文向量：
    
$$context^{<t>} = \sum_{t' = 1}^{T_x} \alpha^{<t,t'>}a^{<t'>}\tag{1}$$ 

##### 澄清“上下文”和“c”
- 在讲座视频中，上下文表示为 $c^{\langle t \rangle}$
- 在作业中，我们调用上下文 $context^{\langle t \rangle}$。
    - 这是为了避免与注意力后 LSTM 的内部存储单元变量混淆，该变量也表示为 $c^{\langle t \rangle}$。

### 练习 2 - modelf

按照图 1 和说明中的说明实现 `modelf()`：

* `modelf` 首先通过 Bi-LSTM 运行输入以获得 $[a^{<1>},a^{<2>}, ..., a^{<T_x>}]$。 
* 然后，`modelf` 使用 `for` 循环调用 `one_step_attention()` $T_y$ 次。  在此循环的每次迭代中：
    - 它将计算出的上下文向量 $context^{<t>}$ 提供给注意力后 LSTM。
    - 它通过 softmax 激活值的全连接层运行注意力后 LSTM 的输出。
    - softmax 生成预测 $\hat{y}^{<t>}$。
    
同样，我们定义了全局层，它们将共享 `modelf()` 中使用的权重。

````python
### 你无法编辑此单元格

n_a = 32 # 注意力前、双向 LSTM 的隐藏状态 'a' 的单元数量
n_s = 64 # 注意力后 LSTM 的隐藏状态 "s" 的单元数

# 请注意，这是帖子关注LSTM单元。  
post_activation_LSTM_cell = LSTM(n_s, return_state = True) # 请不要修改此全局变量。
output_layer = Dense(len(machine_vocab), activation=softmax)
````

现在你可以在 `for` 循环中使用这些层 $T_y$ 次来生成输出，并且它们的参数不会被重新初始化。你必须执行以下步骤： 

1. 将输入 `X` 传播到双向 LSTM。
    * [双向](https://keras.io/layers/wrappers/#bidirectional) 
    * [LSTM](https://keras.io/layers/recurrent/#lstm)
    * 请记住，我们希望 LSTM 返回完整序列，而不仅仅是最后一个隐藏状态。  
    
示例代码：

```Python
sequence_of_hidden_states = Bidirectional(LSTM(units=..., return_sequences=...))(the_input_X)
```
    
2. 迭代 $t = 0, \cdots, T_y-1$： 
    1. 调用`one_step_attention()`，传入注意力前双向LSTM中的隐藏状态 $[a^{\langle 1 \rangle},a^{\langle 2 \rangle}, ..., a^{ \langle T_x \rangle}]$序列，以及注意力后LSTM中的前一个隐藏状态 $s^{<t-1>}$进行计算上下文向量 $context^{<t>}$。
    2. 将 $context^{<t>}$ 赋予注意力后 LSTM 单元。 
        - 记得传入之前的隐藏状态$s^{\langle t-1\rangle}$和这个LSTM的单元状态$c^{\langle t-1\rangle}$ 
        * 这将输出新的隐藏状态 $s^{<t>}$ 和新的单元状态 $c^{<t>}$。  

        示例代码：
        ```Python
        _, next_hidden_state, next_cell_state = 
            post_activation_LSTM_cell(inputs=..., initial_state=[prev_hidden_state, prev_cell_state])
        ```   
 请注意，该层实际上是“注意力后 LSTM 单元”。  为了通过自动评分器的目的，请不要修改该全局变量的命名。  当我们向自动评分器部署更新时，此问题将得到解决。
    3. 将 $s^{<t>}$ 传入带 softmax 激活函数的全连接层，得到输出。

 4. 通过将输出添加到输出列表来保存输出。

````python
### 你无法编辑此单元格

# 单元测试 2

def modelf_states_test(target):
    Tx = 30
    n_a = 32
    n_s = 64
    len_human_vocab = 37
    len_machine_vocab = 11

    model = target(Tx, Ty, n_a, n_s, len_human_vocab, len_machine_vocab)

    # 创建测试输入
    X_test = np.random.rand(1, Tx, len_human_vocab)
    s0_test = np.zeros((1, n_s))
    c0_test = np.zeros((1, n_s))

    @tf.function(input_signature=[
        tf.TensorSpec(shape=[None, Tx, len_human_vocab], dtype=tf.float32),
        tf.TensorSpec(shape=[None, n_s], dtype=tf.float32),
        tf.TensorSpec(shape=[None, n_s], dtype=tf.float32)
    ])
    def predict_function(X, s0, c0):
        # 使用输入张量直接调用模型
        return model([X, s0, c0])  

    # 获取模型前五个时间步的输出
    outputs = predict_function(X_test, s0_test, c0_test)

    # 从每个时间步的 LSTM 输出中提取隐藏状态
    hidden_states = [np.array(output) for output in outputs]

    # 检查连续的隐藏状态是否不同
    for i in range(len(hidden_states) - 1):
        assert not np.allclose(hidden_states[i], hidden_states[i + 1]), (
            "Consecutive hidden states should be different.\n"
            "Check if the LSTM cell is using the correct previous states.\n"
            "Make sure you are using s and c, and NOT using s0 and c0 in Step 2.B."
        )

    print("\033[32mAll tests passed!\033[0m")

modelf_states_test(modelf)
````
