#### 什么是音乐“值”（values）？（选读）
你可以非正式地将每个“值”视为一个音符，其中包含音高和持续时间。例如，如果你按下特定钢琴键 0.5 秒，则你刚刚弹奏了一个音符。在音乐理论中，“值”实际上比这更复杂——具体来说，它还捕获同时演奏多个音符所需的信息。例如，在演奏一首音乐作品时，你可能会同时按下两个钢琴键（同时演奏多个音符会生成所谓的“和弦”）。但对于本作业，你无需担心乐理细节。

## 2 - 构建模型

现在，你将构建并训练一个学习音乐模式的模型。 
* 该模型采用形状为 $(m, T_x, 90)$ 的输入 X 和形状为 $(T_y, m, 90)$ 的标签 Y。 
* 你将使用 LSTM 和具有 $n_{a} = 64$ 尺寸的隐藏状态。

````python
# 每个 LSTM 单元的隐藏状态的维度数。
n_a = 64 
````

#### 序列生成使用 for 循环
* 如果你正在构建 RNN，其中在测试时预先给出整个输入序列 $x^{\langle 1 \rangle}, x^{\langle 2 \rangle}, \ldots, x^{\langle T_x \rangle}$，则 Keras 具有简单的内置函数来构建模型。 
* 但是，对于**序列生成，在测试时你不会提前知道 $x^{\langle t\rangle}$ 的所有值**。
* 相反，你将使用 $x^{\langle t\rangle} = y^{\langle t-1 \rangle}$ 一次生成一个。 
    * 时间“t”的输入是前一个时间步“t-1”的预测。
* 因此，你需要实现自己的 for 循环来迭代时间步。 

#### 可共享权重
* 函数 `djmodel()` 将使用 for 循环调用 LSTM 层 $T_x$ 次。
* 所有 $T_x$ 副本都具有相同的权重，这一点很重要。 
    - $T_x$ 步骤应该具有未重新初始化的共享权重。
* 引用全局定义的共享层将在每个时间步利用相同的层对象实例。
* 在Keras中实现具有可共享权重的层的关键步骤是： 
1. 定义层对象（为此你将使用全局变量）。
2. 传播输入时调用这些对象。

#### 三种类型的网络层
* 全局变量所需的层对象已定义。  
    * 只需运行下一个单元格即可创建它们！ 
* 请阅读 Keras 文档并理解这些层： 
    - [Reshape()](https://www.tensorflow.org/api_docs/python/tf/keras/layers/Reshape)：将输出重塑为特定形状。
    - [LSTM()](https://www.tensorflow.org/api_docs/python/tf/keras/layers/LSTM)：长短期记忆层
    - [Dense()](https://www.tensorflow.org/api_docs/python/tf/keras/layers/Dense)：常规的全连接神经网络层。

````python
n_values = 90 # 音乐值的数量
reshaper = Reshape((1, n_values))                  # 用于 djmodel() 的步骤 2.B，如下
LSTM_cell = LSTM(n_a, return_state = True)         # 用于步骤 2.C
densor = Dense(n_values, activation='softmax')     # 用于步骤 2.D
````

* `reshaper`、`LSTM_cell` 和 `densor` 是全局定义的层对象，你将使用它们来实现 `djmodel()`。 
* 为了通过这些层之一传播 Keras 张量对象 X，请使用 `layer_object()`。
    - 对于一个输入，使用 `layer_object(X)`
    - 对于多个输入，将输入放入列表中：`layer_object([X1,X2])`

### 练习 1 - djmodel

实现`djmodel()`。

#### 输入（给定）

* `Input()` 层用于定义输入 `X` 以及初始隐藏状态 'a0' 和单元状态 `c0`。
* `shape` 参数采用不包含批量维度 (`m`) 的元组。
    - 例如，
    ```Python
    X = Input(shape=(Tx, n_values)) # X 有 3 个维度而不是 2 个：(m, Tx, n_values)
    ```
    
**步骤 1：输出**

* 创建一个空列表“outputs”以保存 LSTM Cell 在每个时间步的输出。

#### 步骤 2：遍历各个时间步
* $t \in 1, \ldots, T_x$ 的循环：

#### 2A。从 `X` 中选择“t”时间步向量。
* X 的形状为 (m, Tx, n_values)。
* “t”选择的形状应为 (n_values,)。 
* 回想一下，如果你在 NumPy 而不是 Keras 中实现，你将从 3D NumPy 数组中提取切片，如下所示：
```Python
var1 = array1[:,1,:]
```
    
#### 2B。将 `x` 重塑为 (1, n_values)。
* 使用 `reshaper()` 层。  这是一个函数，它将前一层作为其输入参数。

#### 2C。通过`LSTM_cell`的一步运行`x`。

* 使用上一步的隐藏状态 $a$ 和单元状态 $c$ 初始化 `LSTM_cell`。 
* 使用以下格式：
```python
_, next_hidden_state, next_cell_state = LSTM_cell(inputs=input_x, initial_state=[previous_hidden_state, previous_cell_state])
```
 * 为输入、隐藏状态和单元状态选择适当的变量。

#### 2D。全连接层
* 使用 `densor` 通过全连接+softmax 层传播 LSTM 的隐藏状态。 
    
#### 2E。追加输出
* 将输出附加到“输出”列表。

#### 步骤3：循环后，创建模型
* 使用 Keras `Model` 对象创建模型。有两种方法可以实例化 `Model` 对象。一种是通过子类化，这里不会使用它。相反，你将使用高度灵活的函数式 API，你可能还记得本课程早期的作业！使用Functional API，你将从输入开始，然后使用链式层调用指定模型的前向传播，最后根据输入和输出创建模型。

* 指定输入和输出，如下所示：
```Python
model = Model(inputs=[input_x, initial_hidden_state, initial_cell_state], outputs=the_outputs)
```
 * 然后，为输入张量、隐藏状态、单元状态和输出选择适当的变量。
* 请参阅 [Model](https://www.tensorflow.org/api_docs/python/tf/keras/Model) 的文档

````python
# UNQ_C1 (UNIQUE CELL IDENTIFIER, DO NOT EDIT)
# GRADED FUNCTION: djmodel

def djmodel(Tx, LSTM_cell, densor, reshaper):
    """
    实现由 Tx LSTM 单元组成的 djmodel，其中每个单元负责
    用于根据前一个注释和上下文学习以下注释。
    每个单元格都有以下模式： 
            [X_{t}, a_{t-1}, c0_{t-1}] -> RESHAPE() -> LSTM() -> DENSE()
    参数：
        Tx -- 语料库中序列的长度
        LSTM_cell -- LSTM 层实例
        densor -- 全连接层实例
        reshaper -- 重塑层实例
    
    返回：
        model -- 具有输入 [X, a0, c0] 的 Keras 实例模型
    """
    # 获取输入值的形状
    n_values = densor.units
    
    # 获取隐藏状态向量的编号
    n_a = LSTM_cell.units
    
    # 定义输入层并指定形状
    X = Input(shape=(Tx, n_values)) 
    
    # 定义初始隐藏状态 a0 和初始单元状态 c0
    # 使用 `Input`
    a0 = Input(shape=(n_a,), name='a0')
    c0 = Input(shape=(n_a,), name='c0')
    a = a0
    c = c0
    ### START CODE HERE ### 
    # 步骤 1：创建空列表以在迭代时附加输出（约 1 行）
    outputs = None
    
    # 步骤 2：循环 tx
    for t in range(None):
        
        # 步骤2.A：从X中选择第“t”个时间步向量。 
        x = None
        # 步骤2.B：使用重塑器将x重塑为(1, n_values)（≈1行）
        x = None
        # 步骤 2.C：执行 LSTM_cell 的一个步骤
        _, a, c = None
        # 步骤 2.D：将 densor 应用于 LSTM_Cell 的隐藏状态输出
        out = None
        # 步骤 2.E：将输出附加到 "outputs"
        None
        
    # 步骤3：创建模型实例
    model = Model(inputs=[X, a0, c0], outputs=outputs)
    
    ### END CODE HERE ###
    
    return model
````

#### 创建模型对象
* 运行以下单元格来定义你的模型。 
* 你将使用`Tx=30`。 
* 该单元可能需要几秒钟的时间才能运行。 

````python
### 你无法编辑此单元格

model = djmodel(Tx=30, LSTM_cell=LSTM_cell, densor=densor, reshaper=reshaper)
````

````python
opt = Adam(lr=0.01, beta_1=0.9, beta_2=0.999, decay=0.01)

model.compile(optimizer=opt, loss='categorical_crossentropy', metrics=['accuracy'])
````

最后，我们将 `a0` 和 `c0` 初始化，使 LSTM 的初始状态为零。 

````python
m = 60
a0 = np.zeros((m, n_a))
c0 = np.zeros((m, n_a))
````

### 练习 2 - music_inference_model

实现 `music_inference_model()` 对音乐值序列进行采样。

以下是你需要在生成 $T_y$ 输出字符的 for 循环内实现的一些关键步骤： 

**第1步**：创建一个空列表“outputs”以保存LSTM Cell在每个时间步的输出。

**步骤2.A**：使用`LSTM_Cell`（它接受输入层）以及上一步的“`c`”和“`a`”来生成当前步骤的“`c`”和“`a`”。 

```Python
_, next_hidden_state, next_cell_state = LSTM_cell(input_x, initial_state=[previous_hidden_state, previous_cell_state])
```
 - 为 `input_x`、`hidden_state` 和 `cell_state` 选择适当的变量

**2.B**：通过应用 `densor` 计算“`a`”上的 softmax 来计算输出，以获得当前步骤的输出。 

**2.C**：将输出附加到列表`outputs`。

**2.D**：将最后的输出转换为下一个时间步的新输入。你将分 2 个子步骤完成此操作：
- 使用 `tf.math.argmax` 沿最后一个轴获取预测输出最大值的索引。
- 使用 `tf.one_hot` 将索引转换为其 n_values-one-hot 编码。

**2.E**：使用`RepeatVector(1)(x)`将shape=(None, 90)编码的one-hot的输出转换为shape=(None, 1, 90)的张量

#### 步骤3：推理模型：  
这是 Keras `Model` 对象的使用方法：

```Python
model = Model(inputs=[input_x, initial_hidden_state, initial_cell_state], outputs=the_outputs)
```

* 为输入张量、隐藏状态、单元状态和输出选择适当的变量。

**提示**：模型的输入是**初始**输入和状态。

````python
# UNQ_C2 (UNIQUE CELL IDENTIFIER, DO NOT EDIT)
# GRADED FUNCTION: music_inference_model

def music_inference_model(LSTM_cell, densor, Ty=100):
    """
    使用 model() 中经过训练的 "LSTM_cell" 和 "densor" 生成值序列。
    
    参数：
    LSTM_cell -- 来自 model()、Keras 层对象训练的 "LSTM_cell"
    densor -- 来自 model()、Keras 层对象训练的 "densor"
    Ty -- 整数，生成的时间步数
    
    返回：
    inference_model -- Keras 模型实例
    """
    
    # 获取输入值的形状
    n_values = densor.units
    # 获取隐藏状态向量的编号
    n_a = LSTM_cell.units
    
    # 使用形状定义模型的输入 
    x0 = Input(shape=(1, n_values))
    
    
    # 定义s0，初始隐藏状态为解码器LSTM
    a0 = Input(shape=(n_a,), name='a0')
    c0 = Input(shape=(n_a,), name='c0')
    a = a0
    c = c0
    x = x0

    ### START CODE HERE ###
    # 步骤 1：创建一个 "outputs" 的空列表以稍后存储你的预测值（约 1 行）
    outputs = None
    
    # 步骤 2：循环 Ty 并在每个时间步生成一个值
    for t in range(None):
        # 步骤 2.A：执行 LSTM_cell 的一个步骤。使用 "x"，而不是 "x0"（≈1 行）
        _, a, c = None
        
        # 步骤 2.B：将 Dense 层应用到 LSTM_cell 的隐藏状态输出（≈1 行）
        out = None
        # 步骤 2.C：将预测 "out" 附加到 "outputs"。 out.shape = (None, 90)（≈1 行）
        outputs.append(None)
 
        # 步骤 2.D： 
        # 根据"out"选择下一个值，
        # 将 "x" 设置为所选值的 one-hot 表示形式
        # 请参阅上面的说明。
        x = None
        x = None
        # 步骤 2.E： 
        # 使用 RepeatVector(1) 将 x 转换为 shape=(None, 1, 90) 的张量
        x = None
        
    # 步骤 3：使用正确的 "inputs" 和 "outputs" （约 1 行）创建模型实例
    inference_model = None
    
    ### END CODE HERE ###
    
    return inference_model
````

运行下面的单元格来定义你的推理模型。该模型经过硬编码以生成 50 个值。

````python
### 你无法编辑此单元格
inference_model = music_inference_model(LSTM_cell, densor, Ty = 50)
````

### 练习 3 - predict_and_sample

实现`predict_and_sample()`。 

这个函数接受许多参数，包括输入 `x_initializer`、`a_initializer` 和 `c_initializer`。 

为了预测与该输入对应的输出，你需要执行 3 个步骤：


#### 步骤 1：

* 使用推理模型根据给定的输入集预测输出。输出 `pred` 应该是长度为 $T_y$ 的列表，其中每个元素都是形状为 (1, n_values) 的 NumPy 数组。
```Python
inference_model.predict([input_x_init, hidden_state_init, cell_state_init])
```
 * 从该 `predict_and_sample` 函数的输入参数中选择 `predict` 的适当输入参数。
 
#### 步骤 2：

* 将 `pred` 转换为 $T_y$ 索引的 NumPy 数组。 
    * 每个索引都是通过 `pred` 列表中的一个元素的 `argmax` 来计算的。 
    * 使用[numpy.argmax](https://docs.scipy.org/doc/numpy/reference/generated/numpy.argmax.html)。
    * 设置`axis`参数。
        * 请记住，预测的形状是 $(m, T_{y}, n_{values})$

#### 步骤 3： 

* 将索引转换为其 one-hot 向量表示。 
    * 使用[to_categorical](https://www.tensorflow.org/api_docs/python/tf/keras/utils/to_categorical)。
    * 设置`num_classes`参数。请注意，出于评分目的：你需要：
        * 使用 `predict_and_sample()` 给定参数中的维度（例如，x_initializer 的维度之一具有不同类数量的值）。

````python
# UNQ_C3 (UNIQUE CELL IDENTIFIER, DO NOT EDIT)
# GRADED FUNCTION: predict_and_sample

def predict_and_sample(inference_model, x_initializer = x_initializer, a_initializer = a_initializer, 
                       c_initializer = c_initializer):
    """
    使用推理模型预测值的下一个值。
    
    参数：
    inference_model -- Keras 推理时间模型实例
    x_initializer -- NumPy 形状数组 (1, 1, 90)，one-hot 向量初始化值生成
    a_initializer -- NumPy 形状为 (1, n_a) 的数组，初始化 LSTM_cell 的隐藏状态
    c_initializer -- NumPy 形状为 (1, n_a) 的数组，初始化 LSTM_cel 的单元状态
    
    返回：
    results -- 形状为 (Ty, 90) 的数组，表示生成值的 one-hot 向量矩阵
    indices -- NumPy 形状为 (Ty, 1) 的数组，表示生成值的索引矩阵
    """
    
    n_values = x_initializer.shape[2]
    
    ### START CODE HERE ###
    # 步骤 1：使用推理模型预测给定 x_initializer、a_initializer 和 c_initializer 的输出序列。
    pred = inference_model.predict([None, None, None])
    # 步骤 2：将 "pred" 转换为具有最大概率的索引的 np.array()
    indices = None
    # 步骤 3：将索引转换为 one-hot 向量，结果的形状应为 (Ty, n_values)
    results = to_categorical(None, num_classes=None)
    ### END CODE HERE ###
    
    return results, indices
````

### 3.2 - 生成音乐 

终于！你已准备好生成音乐。 

你的 RNN 生成一系列值。以下代码通过首先调用 `predict_and_sample()` 函数来生成音乐。然后，这些值被后处理为音乐和弦（意味着可以同时演奏多个值或音符）。 

大多数计算音乐算法都会使用一些后处理，因为没有后处理就很难生成听起来不错的音乐。后处理会执行一些操作，例如通过确保相同的声音不会重复太多次来清理生成的音频，或者两个连续的音符在音高上彼此距离不太远，等等。 

有人可能会说这些后处理步骤中有很多都是黑客行为；此外，许多音乐生成文献也关注手工制作的后处理器，并且很多输出质量取决于后处理的质量而不仅仅是模型的质量。但这种后处理确实会产生巨大的差异，因此你也应该在实现中使用它。
